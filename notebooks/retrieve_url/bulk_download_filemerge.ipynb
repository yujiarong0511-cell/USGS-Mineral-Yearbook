This module aims to merge by row urls bulk retrieved from USGS website, which are currently stored in seperate files by continent.

```python
import os
import pandas as pd

original_url_folder = '../../data/original_url/'
original_url_files = os.listdir(original_url_folder)
# Initialize merged file
usgs_merged = pd.DataFrame({
    'year': [],
    'country': [],
    'url': [],
    'continent': [] # Add an additional column (continent) to indicate the source file (and corresponding website)
})
# 
for file_name in original_url_files:
    # DEBUG: The second parameter (file name) was incorrectly set as "original_url_files[0]"" (which is a constant not reflecting changes in file_name, thus leading to repetition African countries) 
    file_path = os.path.join(original_url_folder, file_name) # Join folder path with file name (with suffix)
    current_continent = file_name.split('.')[0].split('_')[1] # Extract continent from file name
    
    current_file = pd.read_csv(file_path) # Read-in the file to DataFrame format
    current_file['continent'] = current_continent # Add a new column to DataFrame
    usgs_merged = pd.concat([usgs_merged,current_file], ignore_index=True) # Concatenate current file to merged file by row

# Re-check
usgs_merged # 2063 lines in total
# print(usgs_merged.info())
```

```python
## Output the DataFrame
usgs_merged.to_csv('../../data/merged_url/usgs_merged.csv', index=False)
```

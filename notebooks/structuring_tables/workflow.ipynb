This script compiles all modules for processing individual table.

```python
import os
import sys
sys.path.append("../../modules/structuring_tables/")
from workflow import structuring_tables, cleansing_table, decoding_indent, filling_unit
from openpyxl import load_workbook
```

```python
continent_select = 'africa'
origin_dir = "../../data/original_table/usgs_" +continent_select + "/"
country_2019_list = [name for name in os.listdir(origin_dir) if "_2019.xlsx" in name]
print(country_2019_list)
```

```python
error_list = []
for i in range(44, len(country_2019_list)):
    filename = country_2019_list[i]
    country_year = filename.split(".")[0]
    print(f"{i}: {country_year}")

    # Check if there are at least 3 sheets in the original file
    origin_filepath = os.path.join(origin_dir, filename)
    fullfile_openpyxl = load_workbook(origin_filepath)
    if len(list(fullfile_openpyxl)) < 3:
        print("Table 2 does not exist!")
        error_list.append(i)
    else:
        try:
            structuring_tables(continent_select, country_year)
            cleansing_table(continent_select, country_year)
            decoding_indent(continent_select, country_year)
            filling_unit(continent_select, country_year)
        except Exception as e:
            print(f"File: [{i}] error! Reason: {e}")
            error_list.append(i)
            continue
```

In this script we merge URLs to tables from each continent to a single table.

Note: On July 9, 2026 (as we coded the script), we failed to retrieve Table URLs for European countries from USGS website, thus they are not included in the merged table.

```python
import pandas as pd
import os
```

```python
# Retrieve local paths for all continent-specific URL tables
url_dir = "../../data/original_url/tables_new/"
url_filenames = os.listdir("../../data/original_url/tables_new")
url_filenames
```

```python
# Add individual continent-specific URL table to the output table
merged_url_list = pd.DataFrame(columns=["year","country","table_url","continent"])
for filename in url_filenames:
    continent = filename.strip(".csv").split("_")[1]
    filepath = os.path.join(url_dir, filename)
    # Read-in URL table
    url_list = pd.read_csv(filepath)
    # Add a new column (`continent`) to continent-specific URL table
    url_list['continent'] = continent
    # Add a new column indicating the suffix of URL (xls/xlsx)
    url_list['suffix'] = [st[-1] for st in url_list["table_url"].str.split('.')]
    # Concatenate to output table
    merged_url_list = pd.concat([merged_url_list, url_list])
```

```python
# Write merged table to CSV file
merged_url_list.to_csv("../../data/merged_url/usgs_merged_tables.csv", index=False)
```
